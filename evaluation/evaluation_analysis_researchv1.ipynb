Plotting Graphs

```python
# plot the emotion counts:
import pandas as pd
import matplotlib.pyplot as plt


emotion_counts = responses['Emotion'].value_counts()

emotion_counts.plot(kind='bar', title='Emotion Values by Category For Participants')

plt.xlabel('Emotion')
plt.ylabel('Counts')
plt.xticks(rotation=0)
plt.show()
```

```python
# plot the precision at k counts:
import pandas as pd
import matplotlib.pyplot as plt
emotion_counts = responses['Precision @K '].value_counts()

#index to numeric (to preserve order for better visualization)
emotion_counts.index = pd.to_numeric(emotion_counts.index, errors='coerce')
emotion_counts = emotion_counts.sort_index()

emotion_counts.plot(kind='bar', title='Precision @K Values For Participants')

plt.xlabel('Precision @K')
plt.ylabel('Counts')
plt.xticks(rotation=0)
plt.show()
```
